```python
import chromadb

print("ChromaDB imported successfully!")
print("Version:", chromadb.__version__)
```

```python
client = chromadb.Client()
```

```python
collection = client.create_collection("news")
```

```python
collection.add(
    ids=["id1", "id2", "id3", "id4"],
    documents=[
        "Apple is leading in a smart phone game with iPhone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C, and various antioxidants",
        "SpaceX got NASA contract worth 10 billion $",
    ]
)
```

```python
results = collection.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results=2
)

print(results)
```

```python
results = collection.query(
    query_texts=["I love iPad"],
    n_results=2
)

print(results)
```

```python
results = collection.query(
    query_texts=["Tips for improving health"],
    n_results=2
)

print(results)
```

```python
results = collection.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results=2
)

print("Retrieved Documents:\n")

for document in results["documents"][0]:
    print("-", document)
```

```python
results = collection.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results=2
)

print("Document IDs:")
print(results["ids"][0])

print("\nRetrieved Documents:")
print(results["documents"][0])

print("\nDistances:")
print(results["distances"][0])
```
